## B. Fonctions pour estimer les probabilités

On estime les probabilités par des fréquences, par exemple
$P(B|G) = \dfrac{\text{nb de blonds dans G}}{\text{nb de gens dans G}}$.

```python
## La proportion de personnes vivant dans un pays donné
def proba_pays(liste, pays):
    nb = 0
    for ind in liste:
        if ind[2] == pays:
            nb += 1
    return nb / len(liste)

## Une probabilité conditionnelle  cheveux sachant pays
def proba_cheveux_sachant_pays(liste, cheveux, pays):
    nb_pays = 0
    nb = 0
    for ind in liste:
        if ind[2] == pays:
            nb_pays += 1
            if ind[0] == cheveux:
                nb += 1
    return nb / nb_pays


## Une probabilité conditionnelle  cheveux sachant pays
def proba_taille_sachant_pays(liste, taille, pays):
    nb_pays = 0
    nb = 0
    for ind in liste:
        if ind[2] == pays:
            nb_pays += 1
            if ind[1] == taille:
                nb += 1
    return nb / nb_pays

##  La proportion de personnes ayant à la fois une couleur de cheveux ET une taille spécifique, au sein d'un pays précis
def proba_BT_sachant_pays(liste, cheveux, taille, pays):
    nb_pays = 0
    nb = 0
    for ind in liste:
        if ind[2] == pays:
            nb_pays += 1
            if ind[0] == cheveux and ind[1] == taille:
                nb += 1
    return nb / nb_pays

## La proportion globale de personnes ayant cette combinaison de cheveux et de taille dans tout le jeu de données
def proba_BT(liste, cheveux, taille):
    nb = 0
    for ind in liste:
        if ind[0] == cheveux and ind[1] == taille:
            nb += 1
    return nb / len(liste)
```

## C. Maximum a posteriori

On veut comparer $P(G|B,T)$ et $P(P|B,T)$ avec la formule de Bayes :
$$P(G|B,T) = \frac{P(B,T|G)P(G)}{P(B,T)}$$

### C.1 Avec Bayes naïf

On suppose que cheveux et taille sont indépendants sachant le pays : $P(B,T|G) = P(B|G)P(T|G)$.
Pour $P(B,T)$ on utilise la formule des probabilités totales : $P(B,T) = P(B,T|G)P(G) + P(B,T|P)P(P)$.

```python
def map_naif(liste):
    haut_G = proba_cheveux_sachant_pays(liste, "B", "G") * proba_taille_sachant_pays(liste, "T", "G") * proba_pays(liste, "G")
    haut_P = proba_cheveux_sachant_pays(liste, "B", "P") * proba_taille_sachant_pays(liste, "T", "P") * proba_pays(liste, "P")
    pBT = haut_G + haut_P
    return haut_G / pBT, haut_P / pBT

pG, pP = map_naif(individus)
print("P(G|B,T) =", round(pG, 3))
print("P(P|B,T) =", round(pP, 3))
if pG > pP:
    print("MAP naif : Greenland")
elif pP > pG:
    print("MAP naif : Poland")
else:
    print("egalite")
```

### C.2 Sans Bayes naïf

Cette fois on estime directement $P(B,T|G)$ en comptant les gens de G qui sont blonds ET grands.
$P(B,T)$ est estimée par la proportion de blonds grands dans tout le tableau.

```python
def map_non_naif(liste):
    pBT = proba_BT(liste, "B", "T")
    if pBT == 0:
        print("pas de blond grand dans les donnees")
        return 0, 0
    pG = proba_BT_sachant_pays(liste, "B", "T", "G") * proba_pays(liste, "G") / pBT
    pP = proba_BT_sachant_pays(liste, "B", "T", "P") * proba_pays(liste, "P") / pBT
    return pG, pP

pG, pP = map_non_naif(individus)
print("P(G|B,T) =", round(pG, 3))
print("P(P|B,T) =", round(pP, 3))
if pG > pP:
    print("MAP non naif : Greenland")
elif pP > pG:
    print("MAP non naif : Poland")
else:
    print("egalite")
```

Remarque : ici on a en fait $P(G|B,T) = \dfrac{\text{nb de blonds grands de G}}{\text{nb de blonds grands}}$.

## D. Maximum de vraisemblance

Pour le MLE on ne prend pas en compte $P(G)$ et $P(P)$, on compare juste $P(B,T|G)$ et $P(B,T|P)$.

### D.1 Avec Bayes naïf

```python
def mle_naif(liste):
    lG = proba_cheveux_sachant_pays(liste, "B", "G") * proba_taille_sachant_pays(liste, "T", "G")
    lP = proba_cheveux_sachant_pays(liste, "B", "P") * proba_taille_sachant_pays(liste, "T", "P")
    return lG, lP

lG, lP = mle_naif(individus)
print("P(B,T|G) =", round(lG, 3))
print("P(B,T|P) =", round(lP, 3))
if lG > lP:
    print("MLE naif : Greenland")
elif lP > lG:
    print("MLE naif : Poland")
else:
    print("egalite")
```

### D.2 Sans Bayes naïf

```python
def mle_non_naif(liste):
    lG = proba_BT_sachant_pays(liste, "B", "T", "G")
    lP = proba_BT_sachant_pays(liste, "B", "T", "P")
    return lG, lP

lG, lP = mle_non_naif(individus)
print("P(B,T|G) =", round(lG, 3))
print("P(B,T|P) =", round(lP, 3))
if lG > lP:
    print("MLE non naif : Greenland")
elif lP > lG:
    print("MLE non naif : Poland")
else:
    print("egalite")
```

Sur notre tableau, le MAP (naïf et non naïf) donne Greenland alors que le MLE (naïf et non naïf) donne Poland.


## E. Point de vue continu

### E.1 Génération

Maintenant la couleur des cheveux est une intensité entre 0 (blond) et 1 (dark). On la tire uniformément sur (0,1).

```python
def generer_donnees_continues(n):
    tableau = []
    for i in range(n):
        ligne = []
        for j in range(3):
            intensite = round(random.random(), 3)
            taille = random.choice(["T", "S"])
            if j < 2:
                pays = "G"
            else:
                pays = "P"
            ligne.append((intensite, taille, pays))
        tableau.append(ligne)
    return tableau

data_c = generer_donnees_continues(50)
for ligne in data_c[:8]:
    print(ligne)
```

### E.2 Comment répondre à la question

Un individu blond a maintenant une intensité faible, on prend par exemple $x_0 = 0.2$.

Une première idée serait de couper à 0.5 (en dessous = B, au dessus = D) et de refaire exactement comme avant,
mais on perd de l'information (0.01 et 0.49 deviennent pareil).

Une meilleure idée : on garde la formule de Bayes mais on remplace $P(B|G)$ par une densité $f(x_0|G)$ :
$$P(G|x_0,T) = \frac{f(x_0|G)\,P(T|G)\,P(G)}{f(x_0|G)P(T|G)P(G) + f(x_0|P)P(T|P)P(P)}$$
(version naïve). Pour la version non naïve on utilise $f(x_0|T,G)\,P(T|G)$, c'est-à-dire la densité de l'intensité
seulement chez les grands de G.

Pour estimer la densité on suppose qu'elle est gaussienne et on estime la moyenne et l'écart type
à partir des données. (Ce n'est pas parfait car l'intensité est entre 0 et 1, mais c'est le plus simple.)

### E.3 Modifications des fonctions

- les fonctions `proba_pays` et `proba_taille_sachant_pays` ne changent pas ;
- `proba_cheveux_sachant_pays` est remplacée par une fonction qui calcule une densité gaussienne en $x_0$ ;
- pour le non naïf, `proba_BT_sachant_pays` devient densité de l'intensité chez les grands du pays fois $P(T|pays)$ ;
- le reste (comparaison, MAP, MLE) est pareil.

```python
def densite_intensite(liste, x0, pays, taille=None):
    valeurs = []
    for ind in liste:
        if ind[2] == pays:
            if taille == None or ind[1] == taille:
                valeurs.append(ind[0])
    m = np.mean(valeurs)
    s = np.std(valeurs)
    return np.exp(-(x0 - m)**2 / (2 * s**2)) / (s * np.sqrt(2 * np.pi))

def map_naif_continu(liste, x0):
    haut_G = densite_intensite(liste, x0, "G") * proba_taille_sachant_pays(liste, "T", "G") * proba_pays(liste, "G")
    haut_P = densite_intensite(liste, x0, "P") * proba_taille_sachant_pays(liste, "T", "P") * proba_pays(liste, "P")
    total = haut_G + haut_P
    return haut_G / total, haut_P / total

def map_non_naif_continu(liste, x0):
    haut_G = densite_intensite(liste, x0, "G", "T") * proba_taille_sachant_pays(liste, "T", "G") * proba_pays(liste, "G")
    haut_P = densite_intensite(liste, x0, "P", "T") * proba_taille_sachant_pays(liste, "T", "P") * proba_pays(liste, "P")
    total = haut_G + haut_P
    return haut_G / total, haut_P / total

def mle_naif_continu(liste, x0):
    lG = densite_intensite(liste, x0, "G") * proba_taille_sachant_pays(liste, "T", "G")
    lP = densite_intensite(liste, x0, "P") * proba_taille_sachant_pays(liste, "T", "P")
    return lG, lP

def mle_non_naif_continu(liste, x0):
    lG = densite_intensite(liste, x0, "G", "T") * proba_taille_sachant_pays(liste, "T", "G")
    lP = densite_intensite(liste, x0, "P", "T") * proba_taille_sachant_pays(liste, "T", "P")
    return lG, lP

def afficher_resultats(liste, x0):
    print("x0 =", x0)
    noms = ["MAP naif", "MAP non naif", "MLE naif", "MLE non naif"]
    fonctions = [map_naif_continu, map_non_naif_continu, mle_naif_continu, mle_non_naif_continu]
    for i in range(4):
        a, b = fonctions[i](liste, x0)
        if a > b:
            rep = "G"
        else:
            rep = "P"
        print(noms[i], ":", round(a, 3), round(b, 3), "->", rep)

ind_c = mettre_en_liste(data_c)
afficher_resultats(ind_c, 0.2)
```

Comme avant l'intensité ne dépend pas du pays dans nos données, donc le MAP est surtout décidé par $P(G)=2/3$
et le MLE c'est un peu du hasard.

Pour tester si nos fonctions marchent vraiment, on fait un autre jeu de données où les gens de G sont plutôt blonds
(intensité entre 0 et 0.6) et ceux de P plutôt bruns (entre 0.4 et 1).données où les gens de G sont plutôt blonds
(intensité entre 0 et 0.6) et ceux de P plutôt bruns (entre 0.4 et 1).

```python
def generer_donnees_test(n):
    tableau = []
    for i in range(n):
        ligne = []
        for j in range(3):
            taille = random.choice(["T", "S"])
            if j < 2:
                ligne.append((random.uniform(0, 0.6), taille, "G"))
            else:
                ligne.append((random.uniform(0.4, 1), taille, "P"))
        tableau.append(ligne)
    return tableau

ind_test = mettre_en_liste(generer_donnees_test(50))
afficher_resultats(ind_test, 0.2)
print()
afficher_resultats(ind_test, 0.9)
```

Là ça marche : pour $x_0 = 0.2$ toutes les méthodes disent G et pour $x_0 = 0.9$ elles disent P.
